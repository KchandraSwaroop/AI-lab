```python

def unify(x, y, subs=None):
    if subs is None:
        subs = {}

    if x == y:
        return subs

    if isinstance(x, str) and x.islower():
        subs[x] = y
        return subs

    if isinstance(y, str) and y.islower():
        subs[y] = x
        return subs

    return None


# User inputs
psi1 = input("Enter Psi1:")
psi2 = input("Enter Psi2: ")

# Extract arguments
p1 = psi1[psi1.index("(") + 1:psi1.index(")")].split(",")
p2 = psi2[psi2.index("(") + 1:psi2.index(")")].split(",")

# Check predicate and arguments
if psi1.split("(")[0] != psi2.split("(")[0] or len(p1) != len(p2):
    print("Unification Failed")
else:
    result = {}

    for a, b in zip(p1, p2):
        a, b = a.strip(), b.strip()
        result = unify(a, b, result)

        if result is None:
            break

    if result is not None:
        print("Unification Successful")
        print("Substitution:", result)
    else:
        print("Unification Failed")
```
