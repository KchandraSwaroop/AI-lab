```python
KB = [
    ["~Eats(Anil,Peanuts)", "~NotKilled(Anil)", "Food(Peanuts)"],
    ["~Alive(Anil)", "NotKilled(Anil)"],
    ["~Food(Peanuts)", "Likes(John,Peanuts)"],
    ["Alive(Anil)"],
    ["Eats(Anil,Peanuts)"],
    ["~NotKilled(Anil)", "Alive(Anil)"]
]

KB.append(["~Likes(John,Peanuts)"])

def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:

            if l1 == "~" + l2 or l2 == "~" + l1:
                return [
                    x for x in c1 if x != l1
                ] + [
                    x for x in c2 if x != l2
                ]

    return None

print("NAME - K Chandra Swaroop")
print("USN - 1WN24CS128")

r1 = resolve(KB[1], KB[3])
print("Resolvent 1:", r1)

r2 = resolve(KB[0], KB[4])
print("Resolvent 2:", r2)

r3 = resolve(r2, r1)
print("Resolvent 3:", r3)

r4 = resolve(KB[2], r3)
print("Resolvent 4:", r4)

r5 = resolve(r4, KB[6])
print("Final resolvent:", r5)

if r5 == []:
    print("\nGoal proved: John likes peanuts.")
```
